```python
# Tirando algumas colunas do res43, e fazendo a junção deles, juntamente com a criação de uma coluna para identificar se é essencial, estrategico ou normal

df_res43_CEssencial = df_res43_essencial.drop(columns = ['Seq.','Código BEC','Classe BEC','Descrição','Centro de Controle'])
df_res43_CEstrategico = df_res43_estrategico.drop(columns = ['Seq.','Código BEC','Classe BEC','Descrição','Centro de Controle'])
df_res43_CNormal = df_res43_normal.drop(columns = ['Seq.','Código BEC','Classe BEC','Descrição','Centro de Controle'])


# Código para mesclar as três bases e criar uma coluna para Identificar sua classificação

df_res43_CEssencial['Classificação RES43'] = 'Essencial'
df_res43_CEstrategico['Classificação RES43'] = 'Estratégico'
df_res43_CNormal['Classificação RES43'] = 'Normal'
df_res43_mesclado = pd.concat([df_res43_CEssencial, df_res43_CEstrategico, df_res43_CNormal], ignore_index=True)

# Exibindo informações e as primeiras linhas para conferir
display(df_res43_mesclado.head())
print(df_res43_mesclado['Classificação RES43'].value_counts())
```

###**Finalizando com RES, RES82**


---



```python
# Apagando as colunas todas com valor 0

df_res82_final = df_res82.drop(columns = ['Difal ICMS','IPI','Glosa','Percentual','Valor'])
df_res82_final = df_res82_final.fillna('Sem Informação')
df_res82_final = df_res82_final.replace(r'^\s*$', 'Sem Informação', regex=True)
```

```python
# Lista de colunas numéricas do RES82 para padronizar
colunas_valores_res82 = ['Valor Unit.', 'Valor Total']

for col in colunas_valores_res82:
    if col in df_res82_final.columns:
        # Convertendo para string e removendo espaços em branco
        series_str = df_res82_final[col].astype(str).str.strip()

        # Identifica se a coluna usa vírgula como decimal (padrão brasileiro)
        has_comma = series_str.str.contains(',', regex=False)

        # Aplica substituições condicionais de forma segura utilizando np.where e tratando como Series do pandas
        cleaned_series = pd.Series(np.where(
            has_comma,
            series_str.str.replace('.', '', regex=False).str.replace(',', '.', regex=False),
            series_str
        ))

        # Converte para numérico de forma segura, preenchendo falhas/nulos com 0.0
        df_res82_final[col] = pd.to_numeric(cleaned_series, errors='coerce').fillna(0.0)

        # Formata explicitamente para duas casas decimais e garante que o separador decimal seja a vírgula
        df_res82_final[col] = df_res82_final[col].map(lambda val: f"{val:.2f}".replace('.', ','))

# Exibindo o resultado final do DataFrame RES82 para conferência
display(df_res82_final.head())
```

```python
if 'Produto' in df_res82_final.columns:
    # Recarregamos temporariamente a coluna Produto original do df_res82 para garantir um reprocessamento limpo
    df_res82_final['Produto'] = df_res82['Produto'].fillna('Sem Informação').astype(str)

    # 1. Fazemos a primeira divisão para extrair o código estruturado principal
    divisao_codigo = df_res82_final['Produto'].str.split(r'\s*-\s*', n=1, expand=True)

    # O código inicial (ex: '1.03.01.01.00013')
    codigo_principal = divisao_codigo[0].str.strip()
    resto_produto = divisao_codigo[1]

    # Função para extrair o dígito verificador, montar o código completo e limpar o nome do produto
    def extrair_codigo_e_nome(row):
        texto_resto = row[1]
        cod_princ = row[0]

        if pd.isna(texto_resto):
            # Se não houve divisão por hífen, não há dígito verificador separado
            return 'Sem Informação', str(cod_princ).strip()

        # Divide o resto no primeiro hífen
        partes = str(texto_resto).split('-', 1)
        if len(partes) > 1 and partes[0].strip().isalnum() and len(partes[0].strip()) <= 3:
            digito = partes[0].strip()
            nome_produto = partes[1].strip()
            # Junta o código principal com o dígito sem espaços: '1.03.01.01.00013-7'
            codigo_completo = f"{str(cod_princ).strip()}-{digito}"
            return codigo_completo, nome_produto
        else:
            # Caso não siga o padrão de dígito verificador
            return str(cod_princ).strip(), str(texto_resto).strip()

    # Aplicamos a lógica linha por linha
    resultados = divisao_codigo.apply(extrair_codigo_e_nome, axis=1)

    df_res82_final['Código do Prod'] = [r[0] for r in resultados]
    df_res82_final['Produto'] = [r[1] for r in resultados]

    # Reordenando as colunas dinamicamente para posicionar o 'Código do Prod' logo antes de 'Produto'
    colunas = list(df_res82_final.columns)
    if 'Código do Prod' in colunas and 'Produto' in colunas:
        colunas.remove('Código do Prod')
        idx_produto = colunas.index('Produto')
        colunas.insert(idx_produto, 'Código do Prod')
        df_res82_final = df_res82_final[colunas]

# Exibindo as primeiras linhas para conferir o resultado final exatamente como solicitado
display(df_res82_final.head())
```

```python
if 'Centro de Controle' in df_movimentacoes.columns:
    # Garantindo que os dados sejam tratados como string e preenchendo nulos
    df_movimentacoes['Centro de Controle'] = df_movimentacoes['Centro de Controle'].fillna('Sem Informação').astype(str)

    # Divide no primeiro hífen encontrado
    divisao_cc = df_movimentacoes['Centro de Controle'].str.split(r'\s*-\s*', n=1, expand=True)

    # Cria as novas colunas
    df_movimentacoes['Código Centro de Controle'] = divisao_cc[0].str.strip()
    df_movimentacoes['Nome Centro de Controle'] = divisao_cc[1].fillna('Sem Informação').str.strip()

    # Reordenando as colunas para colocar as novas logo após a original
    colunas = list(df_movimentacoes.columns)
    if 'Código Centro de Controle' in colunas and 'Nome Centro de Controle' in colunas:
        colunas.remove('Código Centro de Controle')
        colunas.remove('Nome Centro de Controle')
        idx_original = colunas.index('Centro de Controle')
        colunas.insert(idx_original + 1, 'Código Centro de Controle')
        colunas.insert(idx_original + 2, 'Nome Centro de Controle')
        df_movimentacoes_final = df_movimentacoes[colunas]

    # Removendo a coluna original 'Centro de Controle'
    df_movimentacoes_final = df_movimentacoes_final.drop(columns=['Centro de Controle'])

# Exibindo as colunas modificadas para conferência
display(df_movimentacoes_final[['Código Centro de Controle', 'Nome Centro de Controle']].head())
```
