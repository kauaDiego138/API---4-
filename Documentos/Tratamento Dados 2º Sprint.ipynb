###Criando Padronização de UN Medida

---



```python
#Arquivos com UN Medida são Res34, Res47 e o Res75, primeiro checar qual as unidades presentes em cada um e depois definir cada Unidade

print('Valores únicos de Unid. Medida no df_res34:')
display(df_res34['Unid. Medida'].unique())

print('\nValores únicos de Unid. Medida no df_res47:')
display(df_res47['Unid. Medida'].unique())

print('\nValores únicos de Unid. Medida no df_res75:')
display(df_res75['Unid. Medida'].unique())
```

```python
# Padronizando as unidades de medida nos DataFrames df_res34, df_res47 e df_res75
for df in [df_res34, df_res47, df_res75]:
    if 'Unid. Medida' in df.columns:
        # Substitui 'pç' por 'un' (ignorando maiúsculas/minúsculas antes de padronizar)
        df['Unid. Medida'] = df['Unid. Medida'].astype(str).str.lower().str.replace('pç', 'un', regex=False)
        # Deixa tudo em maiúsculo ('un' vira 'UN')
        df['Unid. Medida'] = df['Unid. Medida'].str.upper().str.strip()

# Verificando o resultado da padronização
print('Unidades pós-padronização no df_res34:', df_res34['Unid. Medida'].unique())
print('Unidades pós-padronização no df_res47:', df_res47['Unid. Medida'].unique())
print('Unidades pós-padronização no df_res75:', df_res75['Unid. Medida'].unique())
```

```python
# Removendo a coluna 'Utilização' do df_base2021
if 'Utilização' in df_base2021.columns:
    df_base2021 = df_base2021.drop(columns=['Utilização'])
    print("Coluna 'Utilização' removida de df_base2021.")
```

```python
def formatar_valores(val):
    if pd.isna(val) or str(val).strip() == '' or str(val).lower() == 'nan':
        return val
    try:
        # Remove espaços
        val_clean = str(val).strip()
        # Tratamento básico para converter string numérica para float
        # Se houver pontos e vírgulas, padroniza
        if ',' in val_clean and '.' in val_clean:
            val_clean = val_clean.replace('.', '').replace(',', '.')
        elif ',' in val_clean:
            val_clean = val_clean.replace(',', '.')

        # Converte para float, formata com 2 casas decimais e altera o ponto por vírgula
        num = float(val_clean)
        return f"{num:.2f}".replace('.', ',')
    except ValueError:
        return val

# Colunas para formatar no df_base2021
colunas_formatar = ['Quantidade', 'Custo Médio', 'Valor Total']

for col in colunas_formatar:
    if col in df_base2021.columns:
        df_base2021[col] = df_base2021[col].apply(formatar_valores)

print("Colunas padronizadas com sucesso no df_base2021!")
display(df_base2021[colunas_formatar].head())
```

```python
# Mapeamento de unidades solicitado para df_base2021 (em minúsculo)
mapeamento_unidades = {
    'fc': 'fl',
    'la': 'l',
    'lt': 'l',
    'mt': 'm',
    'un*': 'un',
    'unid': 'un',
    'rm': 'pct',
    'gr': 'g',
    'em': 'mm',
    'lo': 'un',
    'pç': 'un'
}

# Garantindo a aplicação correta:
# 1. Deixa tudo em minúsculo
# 2. Aplica o mapeamento
# 3. Converte o resultado final para maiúsculo
if 'Unid. Medida' in df_base2021.columns:
    df_base2021['Unid. Medida'] = df_base2021['Unid. Medida'].astype(str).str.lower().str.strip()
    df_base2021['Unid. Medida'] = df_base2021['Unid. Medida'].replace(mapeamento_unidades)
    df_base2021['Unid. Medida'] = df_base2021['Unid. Medida'].str.upper()
    print("Mapeamento e padronização concluídos com sucesso no df_base2021!")
    print("Unidades presentes após mapeamento:", df_base2021['Unid. Medida'].unique())
else:
    print("Coluna 'Unid. Medida' não encontrada no df_base2021.")
```

### Salvando todos os DataFrames na pasta de saída


---



```python
import os
# Criar a pasta de saída se ela não existir
os.makedirs(saida, exist_ok=True)

# Dicionário mapeando os DataFrames para os respectivos nomes de arquivos salvos
dataframes_para_salvar = {
    '0_Estrutura_Organizacional_Tratado2.xlsx': df_est_org,
    '1_PROD_Tratado2.xlsx': df_produtos,
    '2_RES43_Mesclado_Tratado2.xlsx': df_res43,
    '3_Centro_de_Custo_Tratado2.xlsx': df_cc,
    '4_RES34_Tratado2.xlsx': df_res34,
    '5_RES47_Tratado2.xlsx': df_res47,
    '6_RES75_Tratado2.xlsx': df_res75,
    '7_RES82_Tratado2.xlsx': df_res82,
    '8_Movimentacoes_Materiais_Tratado2.xlsx': df_movimentacoes,
    '9_RES47 2021_Tratado.xlsx': df_base2021
}

for nome_arquivo, df_temp in dataframes_para_salvar.items():
    caminho_completo = os.path.join(saida, nome_arquivo)
    df_temp.to_excel(caminho_completo, index=False)
    print(f"Salvo: {nome_arquivo} -> {caminho_completo}")

print("\nTodos os DataFrames foram salvos com sucesso na pasta de saída!")
```
